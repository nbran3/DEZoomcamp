```python
import json
import pandas as pd
from time import time

from kafka import KafkaProducer

def json_serializer(data):
    return json.dumps(data).encode('utf-8')

server = 'localhost:9092'

producer = KafkaProducer(
    bootstrap_servers=[server],
    value_serializer=json_serializer
)

producer.bootstrap_connected()
```

```python
from time import time

t0 = time()

csv_url = "https://github.com/DataTalksClub/nyc-tlc-data/releases/download/green/green_tripdata_2019-10.csv.gz"
df = pd.read_csv(csv_url, compression='gzip')

df = df[['lpep_pickup_datetime', 'lpep_dropoff_datetime', 'PULocationID', 'DOLocationID', 
         'passenger_count', 'trip_distance', 'tip_amount']]

for _, row in df.iterrows():
    message = row.to_dict()  
    producer.send("green_trips", value=message)


producer.flush()
t1 = time()

took = t1 - t0

took
```
